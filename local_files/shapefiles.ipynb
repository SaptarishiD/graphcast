```python
import geopandas as gpd
import matplotlib.pyplot as plt
from shapely.geometry import box

indiashp_path = "/Datastorage/saptarishi.dhanuka_asp25/shapefile_data/india_homog"

# Path to your India shapefile (replace with actual path)
shapefile_path = indiashp_path
regions = ['Central_Northeast', 'Hilly_Regions', 'Northeast', 'Northwest', 'South_Peninsular', 'West_Central']

# Load shapefile and ensure it's in EPSG:4326 (lat/long)
india = gpd.read_file(shapefile_path, layer=regions[2]).to_crs(epsg=4326)
india
```

```python
for region in regions:
    shape =  gpd.read_file(shapefile_path, layer=region).to_crs(epsg=4326)
    fig, axes = plt.subplots(1, 1, figsize=(14, 7))
    gpd.GeoSeries(shape['geometry']).plot(ax=axes, color="lightgreen", edgecolor="black")
    axes.set_title(f"{region}")
    axes.set_xlabel("Longitude")
    axes.set_ylabel("Latitude")
    axes.set_xlim(65, 100)
    axes.set_ylim(5, 37)
    plt.savefig(f'./plots/shapefile_maps/{region}.png')
    plt.show()
    plt.clf()
```
